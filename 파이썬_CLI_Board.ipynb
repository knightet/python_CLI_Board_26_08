```python
#만든 내용 클래스 화
user = input("사용자 입력")
i = 1
class cli:
  def __init__(self): # 게시글 실행
    self.i = 1
    self.title_list = []
    self.main_list = []

  def articlewrite(self): # 게시글 생성
    title = input("제목 : ")
    main = input("내용 : ")
    self.title_list.append(title)
    self.main_list.append(main)
    print("{}번 글이 생성되었습니다.".format(i))
    self.i+=1

  def exit(self): # 게시판 나가는것
    print("게시판 종료")

  def helf(self):# 게시판 명령어 확인
    print("명령어 확인")
    print("article write == 게시판 생성")
    print("exit == 게시판 종료")
    print("article list == 게시판 확인")

  def articlelist(self):
    print("="*30)
    print("{}  /   {}   /   {}  ".format("번호","제목","내용"))
    for j in range(0,self.i-1):
      print("{}    /    {}    /    {}   ".format(j+1,self.title_list[j],self.main_list[j]))
    print("="*30)



board = cli()

print("환영합니다.{}".format(user))
print("== CLI 게시판 실행 ==")
while(1):
  s = input("명령어 : ")
  if (s == "article write"):
    board.articlewrite()
  elif (s == "exit"):
    board.exit()
    print("== CLI 게시판 종료 ==")
    break
  elif (s == "helf"):
    board.helf()
  elif (s == "article list"):
    board.articlelist()
  else :
    print("사용할수 없는 명령어 입니다. 명령어 확인은 helf")
```

```python
#기본적인 내용 만들기
i = 1 #몇번째 숫자인지 출력r
articles = []
article_id =1
print("== CLI 게시판 실행 ==")
while(1):
  s = input("명령어 : ")
  if(s == "article write"):
    title = input("제목 : ")
    main = input("내용 : ")
    article = {
            "id": article_id,
            "title": title,
            "main": main
        }
    articles.append(article)
    print("{}번 글이 생성되었습니다.".format(article_id))
    article_id+=1

  elif(s == "help"): #명령어 도움말
    print("article write == 게시판 생성")
    print("exit == 게시판 종료")
    print("article list == 게시판 확인")
    print("article delete == 게시판 확인")

  elif(s == "exit"):# 프로그램 종료 명령어
    break

  elif (s == "article list"): #게시글 목록 보기
    if not articles:
        print("등록된 게시글이 없습니다.")
        continue
    print("="*30)
    print("{}  /   {}   /   {}  ".format("번호","제목","내용"))
    for article in reversed(articles):
       print("{}    /    {}    /    {}   ".format(article["id"], article["title"], article["main"]))
    print("="*30)

  elif (s == "article delete"):
    if not articles:
            print("삭제할 게시글이 없습니다.")
            continue

    target_id = int(input("몇 번 게시글을 삭제할까요?: "))

        # 입력받은 번호(id)와 일치하는 게시글 찾기
    target_article = None
    for article in articles:
      if article["id"] == target_id:
          target_article = article
          break

    if target_article:
            articles.remove(target_article)
            print("{}번 게시글이 삭제되었습니다.".format(target_id))
    else:
            print("{}번 게시글이 존재하지 않습니다.".format(target_id))
  elif (s == "article edit"): #게시글 수정
    target_id = int(input("몇 번 게시글을 수정할까요?: "))
    target_article = None
    if not articles:
            print("삭제할 게시글이 없습니다.")
            continue
    for article in articles:
      if article["id"] == target_id:
          target_article = article
          break
    if target_article:
            print("{}번 게시글 수정내용을 입력해주세요.".format(target_id))
    else:
            print("{}번 게시글이 존재하지 않습니다.".format(target_id))
            continue
    edit_title = input("제목 : ")
    edit_main = input("내용 : ")
    target_article["title"] = edit_title
    target_article["main"] = edit_main
  elif(s == "article detail"):
    target_id = int(input("몇 번 게시글을 확인할까요?: "))


  else:
    print("입력된 명령어가 없습니다. 명령어 확인 help")


print("== CLI 게시판 종료 ==")
```
