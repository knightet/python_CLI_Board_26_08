```python
#만든 내용 클래스 화
user = input("사용자 입력")
i = 1
class cli:
  def __init__(self): # 게시글 실행
    self.i = 1
    self.title_list = []
    self.main_list = []

  def articlewrite(self): # 게시글 생성
    title = input("제목 : ")
    main = input("내용 : ")
    self.title_list.append(title)
    self.main_list.append(main)
    print("{}번 글이 생성되었습니다.".format(i))
    self.i+=1

  def exit(self): # 게시판 나가는것
    print("게시판 종료")

  def helf(self):# 게시판 명령어 확인
    print("명령어 확인")
    print("article write == 게시판 생성")
    print("exit == 게시판 종료")
    print("article list == 게시판 확인")

  def articlelist(self):
    print("="*30)
    print("{}  /   {}   /   {}  ".format("번호","제목","내용"))
    for j in range(0,self.i-1):
      print("{}    /    {}    /    {}   ".format(j+1,self.title_list[j],self.main_list[j]))
    print("="*30)



board = cli()

print("환영합니다.{}".format(user))
print("== CLI 게시판 실행 ==")
while(1):
  s = input("명령어 : ")
  if (s == "article write"):
    board.articlewrite()
  elif (s == "exit"):
    board.exit()
    print("== CLI 게시판 종료 ==")
    break
  elif (s == "helf"):
    board.helf()
  elif (s == "article list"):
    board.articlelist()
  else :
    print("사용할수 없는 명령어 입니다. 명령어 확인은 helf")
```

```python
#기본적인 내용 만들기
i = 1 #몇번째 숫자인지 출력r
title_list = []
main_list = []
print("== CLI 게시판 실행 ==")
while(1):
  s = input("명령어 : ")
  if(s == "article write"):
    title = input("제목 :")
    main = input("내용 : ")
    title_list.append(title)
    main_list.append(main)
    print("{}번 글이 생성되었습니다.".format(i))
    i+=1
  elif(s == "check"):
    print("article write == 게시판 생성")
    print("exit == 게시판 종료")
    print("article list == 게시판 확인")
  elif(s == "exit"):
    break
  elif (s == "article list"):
    print("="*30)
    print("{}  /   {}   /   {}  ".format("번호","제목","내용"))
    for j in range(i-2,-1,-1):
       print("{}    /    {}    /    {}   ".format(j+1,title_list[j],main_list[j]))
    print("="*30)
  else:
    print("사용할수 없는 명령어 입니다. 명령어 확인은 check")
print("== CLI 게시판 종료 ==")
```
